# Phase 0: reproducible foundations
Synthetic data only. Run from the repository root using the Phase 0 environment. This notebook verifies development tools, not racing performance.

```python
from dataclasses import dataclass
import sys
from pathlib import Path
from shibata.phase0 import make_example, exercise_operations
from shibata.phase0_validation import check_later_period

@dataclass
class StudyRun:
    name: str
    synthetic: bool = True

def describe(run):
    if not run.synthetic:
        raise ValueError("This exercise only accepts synthetic data")
    return {"name": run.name, "synthetic": run.synthetic}

runs = [StudyRun("development"), StudyRun("later-period")]
for run in runs:
    print(describe(run))
try:
    describe(StudyRun("invalid", False))
except ValueError as error:
    print("Expected rejection:", error)
print("Kernel executable:", sys.executable)
```

## CSV / Parquet, grouping and past-only history
The command-line acceptance run verifies CSV/Parquet round trips and SQLite equality. Here, inspect the same transformations. Missing first-start history stays missing. Whole-period wins are descriptive only.

```python
rows = make_example()
transformed, summary = exercise_operations(rows)
display(transformed)
display(summary)
```

## Later-period validation
The unchanged processing function is checked on a separate synthetic 2025 period against manually specified values. Changing current/future labels must not change the history available at each boundary.

```python
later, validation = check_later_period()
print(validation)
display(later)
```

## Plotting
This chart is an exercise visualization, not a performance measurement.

```python
%matplotlib inline
from matplotlib import pyplot as plt
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(summary.horse_id, summary.wins)
ax.set(xlabel="Synthetic horse", ylabel="Total wins (descriptive)", title="Phase 0 plotting exercise")
plt.show()
```
